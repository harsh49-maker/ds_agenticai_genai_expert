```python
import re

# 1. Read your train.py file
with open("train.py", "r") as f:
    content = f.read()

# 2. Automatically update the path to climb back up to where the CSV actually lives
# This tells python to look up two folders and inside 'data' to find your dataset
updated_content = re.sub(r'csv_path = .+', 'csv_path = "../../data/vendor_invoice.csv"', content)

# 3. Save the corrected script
with open("train.py", "w") as f:
    f.write(updated_content)

print("✅ train.py successfully updated to use the correct relative Windows directory path!")
```
