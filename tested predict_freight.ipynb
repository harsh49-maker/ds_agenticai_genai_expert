```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Load your local CSV dataset directly
df = pd.read_csv("vendor_invoice.csv")

# 2. Extract features and train a fresh model instantly in memory
X = df[['Dollars']]
y = df['Freight']

model = LinearRegression()
model.fit(X, y)

# 3. Create the test data and generate predictions instantly
sample_data = {
    "Dollars":[18500,9000,3000,200]
}

input_df = pd.DataFrame(sample_data)
input_df['Predicted_Freight'] = model.predict(input_df[['Dollars']]).round()

print("🎯 Inference Successful! Final output table below:")
print(input_df.to_string(index=False))
```
